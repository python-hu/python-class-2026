### 演習6-1：うるう年の判定

変数 `year` に西暦（正の整数）が格納されているとする．
次の規則に従って，`year` がうるう年であれば「うるう年です」，そうでなければ「うるう年ではありません」と表示するプログラムを作成せよ．

- `4` で割り切れる年はうるう年である
- ただし，`100` で割り切れる年はうるう年ではない
- ただし，`400` で割り切れる年はうるう年である

例えば，2024年はうるう年，2026年はうるう年ではない．また，1900年はうるう年ではなく，2000年はうるう年である．

なお，変数 `year` には適当な値を代入し，プログラムの実行結果も表示せよ．

````python
# 解答例①
# 規則を上から順番に判定する書き方．条件式は上から順番に判定されることを利用している．
year = 2000

if year % 400 == 0:
    print("うるう年です")
elif year % 100 == 0:
    print("うるう年ではありません")
elif year % 4 == 0:
    print("うるう年です")
else:
    print("うるう年ではありません")
````

````python
# 解答例②
# 論理演算子を用いて，1つの条件式にまとめた書き方．
year = 2000

if (year % 4 == 0 and year % 100 != 0) or year % 400 == 0:
    print("うるう年です")
else:
    print("うるう年ではありません")
````

```{admonition} 解答の解説

解答例①では，「例外の例外」である `400` で割り切れるかどうかを最初に判定している点に注意しよう．
`year % 4 == 0` を最初に判定してしまうと，1900年のように `100` で割り切れる年も「うるう年です」と判定されてしまう（第4回の「よくある間違い」で扱った，`elif` の順番の問題と同じである）．

解答例②では，「`4` で割り切れて `100` で割り切れない」または「`400` で割り切れる」という1つの条件式で表現している．
`and` と `or` を組み合わせる場合は，括弧 `()` を用いて判定のまとまりを明示しておくと読みやすい．
```

### 演習6-2：回文の判定

前から読んでも後ろから読んでも同じになる文字列を **回文** という（例：`"level"`，`"しんぶんし"`）．
変数 `word` に文字列が格納されているとする．
`word` が回文であれば「回文です」，そうでなければ「回文ではありません」と表示するプログラムを作成せよ．

なお，変数 `word` には適当な文字列を代入し，プログラムの実行結果も表示せよ．

````python
# 解答例
word = "level"

if word == word[::-1]:
    print("回文です")
else:
    print("回文ではありません")
````

```{admonition} 解答の解説

おそらく躓きやすいポイントとしては，「回文かどうか」を条件式としてどのように表現するか，であろう．

上記の解答例では，第2回で学んだ スライス `[::-1]` を用いて文字列を逆順にし，「元の文字列と，逆順にした文字列が等しいか？」を条件式として表現している．

例えば `word = "level"` の場合，`word[::-1]` も `"level"` となるため条件式は `True` となる．
一方，`word = "python"` の場合，`word[::-1]` は `"nohtyp"` となるため条件式は `False` となる．
```

### 演習6-3：じゃんけんの判定

変数 `me` と `you` に，それぞれ自分と相手の手（`"グー"`，`"チョキ"`，`"パー"` のいずれか）が格納されているとする．
じゃんけんの結果として，自分から見て「勝ち」「負け」「あいこ」のいずれかを表示するプログラムを作成せよ．

**ヒント:** まず「あいこ」かどうかを判定し，その後で「自分が勝つ組み合わせ」を考えよう．自分が勝つ組み合わせは3通りしかない．

なお，変数 `me` と `you` には適当な値を代入し，プログラムの実行結果も表示せよ．

````python
# 解答例
me = "グー"
you = "チョキ"

if me == you:
    print("あいこ")
elif (me == "グー" and you == "チョキ") or (me == "チョキ" and you == "パー") or (me == "パー" and you == "グー"):
    print("勝ち")
else:
    print("負け")
````
