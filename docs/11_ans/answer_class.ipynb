## 演習1：クラスの作成
読書サークルの会員情報を管理するプログラムを作成したい．クラスを用いて次のインスタンス変数を持つクラス `Member` を作成せよ．

必要なインスタンス変数

- ニックネーム（`nickname`）：`str` 型
- 会員ID（`member_id`）：`str` 型


````python
# 解答

class Member:
    def __init__(self, nickname, member_id):
        self.nickname = nickname
        self.member_id = member_id
````


## 演習3：インスタンスメソッド
演習1で作成したクラスを再定義して，以下の情報を出力するインスタンスメソッド `print_data` を作成せよ．

```
夏目漱石の会員IDはNS112です
森鷗外の会員IDはMO205です
樋口一葉の会員IDはHI318です
```


````python
# 解答

class Member:
    def __init__(self, nickname, member_id):
        self.nickname = nickname
        self.member_id = member_id

    def print_data(self):
        print(f'{self.nickname}の会員IDは{self.member_id}です')

club_members = [
    Member('夏目漱石', 'NS112'),
    Member('森鷗外', 'MO205'),
    Member('樋口一葉', 'HI318'),
]

for member in club_members:
    member.print_data()
````

## 演習4：クラスのまとめ
講義資料に含まれていないあなた自身が考えた独自の機能を持つクラスを定義して，インスタンスメソッドから実装した処理の呼び出しを行うコードを作成せよ．独自であれば**どのような機能でも良い**．また **説明** と書かれたMarkdownセルに実装したクラスの機能について説明せよ．

一からクラスを作成することが難しければ，演習1〜3のクラスをベースにしても良い．

````python
# 解答例

class ReadingMember:
    def __init__(self, nickname, member_id, monthly_goal):
        self.nickname = nickname
        self.member_id = member_id
        self.monthly_goal = monthly_goal
        self.books_read = 0

    def log_books(self, count):
        self.books_read = count + self.books_read

    def progress_label(self):
        ratio = self.books_read / self.monthly_goal
        if ratio >= 1:
            return '目標達成'
        if ratio >= 0.5:
            return '折り返し地点'
        return 'これから'

    def print_progress(self):
        label = self.progress_label()
        print(f'{self.nickname}（ID:{self.member_id}）は{self.books_read}冊読了：{label}')

soseki = ReadingMember('夏目漱石', 'NS112', monthly_goal=6)
ogai = ReadingMember('森鷗外', 'MO205', monthly_goal=8)
ichiyo = ReadingMember('樋口一葉', 'HI318', monthly_goal=5)

soseki.log_books(3)
ogai.log_books(2)
ichiyo.log_books(5)

for reading_member in (soseki, ogai, ichiyo):
    reading_member.print_progress()
````

**説明**

サークル内の読書目標を可視化できるクラス `ReadingMember` を実装した．

- ニックネーム，ID，月間読書目標冊数を保持する．この例では，月間読書目標には1以上の整数を指定する．
- `log_books` で読了冊数を蓄積し，`progress_label` で達成状況を判定する．
- `print_progress` を呼び出すと現在の読了冊数と達成ラベルをまとめて表示できる．

メソッドを呼び出すだけでその月の進捗を共有できるため，演習1〜3で扱った会員情報の応用例になる．
