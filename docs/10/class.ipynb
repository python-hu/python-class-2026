## クラスとは？

![](figs/class_image.png)

**クラス（class）** とはデータの値や処理の振る舞いのルールを記述できる仕組みである．この仕組みを理解するために，図に示すような設計図をイメージされたい．私たちは何かモノを作るとき，モノの情報や作り方を決めた設計図を予め作成し，その設計図をもとにモノを作成する．

例えば，カメラの場合だと，製品名，大きさ，デザイン，レンズ，写真の保存の仕方など，製造するために必要な情報が設計図には記載されている．そして，設計図からカメラを製造し，完成品ができる．クラスはこの一連の手順に含まれるモノの設計図としてイメージすると良い．

このたとえでは，Pythonのクラスをもとに作られる **インスタンス** が完成品にあたる．インスタンスは **オブジェクト** とも呼ばれ，インスタンスを作成することを **インスタンス化** という．詳細は後述する．

また，発展的な内容であるが，カメラの性能をアップデートしたり，デザインを変更した改良品の設計図を作成したい場合もある．Pythonでは，改良前の設計図をベースに新たな設計図を作成する **継承** という仕組みもサポートされている．

クラスの簡単なイメージは以上であるが，厳密な定義や詳細な機能はゼロからはじめるプログラミングの内容を大幅に超えるため，本講義では，クラスがどのような場面で便利なのかという視点から，基本的な機能を紹介する．

## クラスの利用場面

次のようなプログラムを作成する場面を想定されたい．あなたは動物園の管理者で，動物種，名前，飼育場所，画像など，飼育動物の情報をコンピュータで管理する必要がある．

まず初めに，飼育されている犬のポチ，猫のタマについて動物種と名前を管理してみよう．

このような複数のデータを扱う場合，リストが便利であった．ここでは，動物種と名前をそれぞれリストに入れ，同じインデックスの要素が同じ動物に対応するようにする．

````python
species = ['犬', '猫']
names = ['ポチ', 'タマ']
````

`for` 文でインデックスを順番に取り出し，`print()` 関数を使って以下のように出力できる．

````python
for i in range(len(species)):
    print(f'{species[i]}：{names[i]}')
````

出力することができた．続いて，動物種に基づいて鳴き声を出力させることを考える．鳴き声もリストで管理することはできるが，ここでは動物種ごとに鳴き声を返す関数を定義する．

````python
def dog_sound():
    return 'ワンワン'

def cat_sound():
    return 'ニャア'

for i in range(len(species)):
    if species[i] == '犬':
        sound = dog_sound()
    elif species[i] == '猫':
        sound = cat_sound()
        
    print(f'{species[i]}：{names[i]} ... {sound}')
````

`for` 文と条件分岐を組み合わせて実装することができた．ではここから，鳥のピヨの情報と鳴き声を返す関数を追加することを考えよう．以下では，鳥を加えたリストと関数を定義し，条件分岐にも鳥の場合の処理を追加する．

````python
species = ['犬', '猫', '鳥']
names = ['ポチ', 'タマ', 'ピヨ']

def bird_sound():
    return 'チュンチュン'

for i in range(len(species)):
    if species[i] == '犬':
        sound = dog_sound()
    elif species[i] == '猫':
        sound = cat_sound()
    elif species[i] == '鳥':
        sound = bird_sound()
        
    print(f'{species[i]}：{names[i]} ... {sound}')
````

新しい動物種である鳥を追加するために，リストの要素に加えて，関数と条件分岐を追加した．同じ動物種の個体を増やすだけなら，既存の鳴き声の関数を利用できる．

## クラスの作成

![](figs/class.png)

上図では，左側が `Animal` クラス，右側がそのクラスから作られる犬・猫・鳥のインスタンスを表している．図中の右側にも「Animal Class」とあるが，それぞれは同じクラスから作られた別々のインスタンスである．動物種や名前を持たせる方法は，次の節で説明する．

クラスは以下の例のように `class` を使うと定義することができる．

```python
class クラス名:
    def __init__(self):
        初期化処理
```

`class` に加えて，`__init__` と `self` という新たな名前が登場した．簡単に説明すると，`__init__` は，インスタンスを作成するときに自動的に呼び出され，初期化処理を行う **初期化メソッド** である．同じクラスから複数のインスタンスを作る場合は，それぞれを作成するときに呼び出される．

また，`self` は，そのインスタンス自身を示す引数である．**インスタンス変数** と呼ばれる，インスタンスに保存された変数にアクセスするために利用する．詳細は後述する．

これ以外にも，クラスには複数の機能がサポートされており，書き方にはさまざまな流儀がある．まずはコピー&ペーストでいいので使い方に慣れてほしい．

上図に示すような動物を管理するための `Animal` クラスを，まずは初期化時にメッセージを表示するだけの形で定義する．


````python
class Animal:
    def __init__(self):
        print('動物のインスタンスを作成しました')
````

上記のセルでは，クラスが持つルール（ここでは初期化メソッドのみ）を定義した．この段階ではインスタンスを作成していないため，`__init__` 内の処理は実行されず，何も表示されない．

クラスを利用するためには，クラスからデータの実体を作成する必要がある．これを **インスタンス化** といい，クラスから作成されたデータの実体を **インスタンス** または **オブジェクト** という．以下のセルでは，`Animal()` と書くことでインスタンスを作成し，変数 `a` に代入する．


````python
a = Animal()
````

上記のセルを実行すると，「動物のインスタンスを作成しました」と表示される．これは，`Animal()` によってインスタンスを作成するときに，初期化メソッド `__init__` が呼び出され，その中の `print()` が実行されたためである．

`__init__` の定義には `self` があるが，`Animal()` と書くときに自分で `self` を渡す必要はない．作成されるインスタンスが自動的に渡される．

```{admonition} クラスを書き換えたときのセルの実行順序
:class: tip

このノートでは，説明に合わせて `Animal` クラスの定義を少しずつ書き換える．新しい定義のセルを実行した後は，インスタンスを作成するセルも実行されたい．クラスを再定義しただけでは，既に作成したインスタンスは新しい定義に置き換わらない．動作が説明と合わなくなった場合は，ノートの上から順番にセルを実行すると確認しやすい．
```


## インスタンス変数の作成
続いて，`self` を理解するために，`Animal` クラスに動物種を追加してみる．以下のように，`__init__` を書き換えることで実装できる．

```python
class クラス名:
    def __init__(self, variable):
        self.variable = variable
```

初期化処理を行う `__init__` メソッドの引数に `variable` が加えられ，メソッド内部に `self.variable = variable` の処理が追加されている．右辺の `variable` は，関数の回で学んだように，引数として受け取った値を表す．左辺の `self.variable` は，`self` が示すインスタンスの変数である．この代入によって，受け取った値をそのインスタンスに保存する．

このようにインスタンスが持つ変数は **インスタンス変数** と呼ばれ，インスタンスごとにデータを保持できるのである．以下で，動物種を表すインスタンス変数 `species` を持つクラスを定義する．


````python
class Animal:
    def __init__(self, species):
        self.species = species
````

インスタンス変数 `self.species` を作成するクラスを定義できた．続いて，クラスからインスタンスを作成する．

ここで関数のときと同様に，`__init__` メソッドには動物種を示す `species` が引数として加えられているので，インスタンス化のときに動物種を与える必要がある．以下の例では，`'犬'` が `species` に渡され，`self.species` に保存される．


````python
a = Animal('犬')
````

今回は，`__init__` で `print()` を使っていないので，インスタンスを作成しても文字列は出力されない．インスタンス変数を参照するためには，`インスタンス.インスタンス変数名` と書く．以下のセルでは，`a.species` の値を確認する．ノートブックでは，セルの最後に書いた式の値が表示される．


````python
a.species
````

続いて，名前 `name` も追加したクラスを作成する．

````python
class Animal:
    def __init__(self, species, name):
        self.species = species
        self.name = name
````

インスタンス化の際には，`species` と `name` の両方を指定する必要がある．

````python
a = Animal('犬', 'ポチ')
````

作成したインスタンス `a` の変数にアクセスし，動物種と名前を出力する．


````python
print(f'{a.species}の{a.name}です．')
````

以上のように，インスタンス変数を持つクラスを定義することでインスタンスにデータをまとめることができた．

クラスの利点を確認するために，他の動物のインスタンスも作成してみる．

````python
a = Animal('犬', 'ポチ')
b = Animal('猫', 'タマ')
c = Animal('鳥', 'ピヨ')
````

作成したインスタンス `a`，`b`，`c` は，それぞれ `species` と `name` という同じ名前のインスタンス変数を持つが，保存されている値は異なる．アクセスするインスタンスに注意しつつ，以下のセルから確認してほしい．


````python
print(f'{a.species}の{a.name}です．')
print(f'{b.species}の{b.name}です．')
print(f'{c.species}の{c.name}です．')
````

以上のように，動物1頭分のデータをインスタンスにまとめて扱うことができた．動物種と名前を別々のリストで管理していた例と比べて，どの情報がどの動物のものかを確認しやすくなった．

インスタンス変数の値は，作成後に変更することもできる．例えば，`a.name = 'シロ'` とすると，`a` の名前だけが変わり，`b` や `c` の名前は変わらない．

また，インスタンス変数には，数値データや文字列だけでなく，リストや別のクラスから作成されたインスタンスも保存できる．`__init__` メソッドの引数には，関数と同様にデフォルト値を設定でき，インスタンス化の際にはキーワード引数も利用できる．


```{admonition} 引数とインスタンス変数の違い
:class: caution

`__init__` 内部の `variable` と `self.variable` は，それぞれ引数として受け取った値を扱う変数と，インスタンスが持つ変数である．この例では，`self.variable = variable` と代入することで値をインスタンスに保存している．そのため，初期化処理が終わった後も，`インスタンス.variable` と書いて保存した値を参照できる．
```

## インスタンスメソッドの定義
前述のセルでは，作成したインスタンスの変数を参照し，`print()` 関数で出力した．データをインスタンスにまとめることはできたが，表示する処理はインスタンスごとに書いていた．そこで，この処理をクラスが持つ機能として追加する．このように，インスタンスのデータを扱う処理は **インスタンスメソッド** として，以下のように定義できる．

```python
class クラス名:
    def __init__(self, variable):
        self.variable = variable

    def インスタンスメソッド名(self, 引数):
        処理
```

`__init__` と同じ深さのインデントで，新たなメソッド（処理）を定義している．これがインスタンスメソッドである．メソッドの第一引数には，そのインスタンス自身を受け取る `self` を指定することに注意されたい．以下が，前述の情報を表示する処理をインスタンスメソッド `print_data` にまとめた `Animal` クラスである．


````python
class Animal:
    def __init__(self, species, name):
        self.species = species
        self.name = name

    def print_data(self):
        print(f'{self.species}の{self.name}です．')
````

初期化メソッド（`__init__`）では，`self.species = species` と `self.name = name` によって，インスタンスが持つ変数を作成した．今回導入した `print_data` メソッド内でも，`self.species` と `self.name` にアクセスしていることに着目されたい．

`print_data` を呼び出すたびに `species` や `name` を引数として渡さなくても，そのインスタンスに保存した値を使って処理できることがポイントである．関数との比較は，後半の発展部分で説明する．


インスタンスメソッドを呼び出すために，まず，新しく定義したクラスからインスタンスを作成する．


````python
a = Animal('犬', 'ポチ')
b = Animal('猫', 'タマ')
c = Animal('鳥', 'ピヨ')
````

インスタンスメソッドは，`インスタンス名.インスタンスメソッド名(引数)` で呼び出すことができる．`a.print_data()` と書くと，`a` が自動的に `self` に渡されるため，呼び出す側で `self` を引数に書く必要はない．以下の例では，他に渡す引数はないので，括弧の中は空にする．


````python
a.print_data()
b.print_data()
c.print_data()
````

以上より，呼び出す側ではインスタンス変数を一つずつ参照しなくても，メソッド内でそのインスタンスの情報を取り出し，表示する処理を実現できた．

インスタンスメソッドは，第8回で学んだ関数のように `return` で値を返すこともできる．動物の情報をリストとして返すインスタンスメソッド `get_data` を実装してみる．


````python
class Animal:
    def __init__(self, species, name):
        self.species = species
        self.name = name

    def print_data(self):
        print(f'{self.species}の{self.name}です．')

    def get_data(self):
        animal_data = [self.species, self.name]
        return animal_data
````

`get_data` からは，動物種と名前を要素に持つリストが返される．以下では，返されたリストを変数 `animal_data` に代入し，出力する．


````python
a = Animal('犬', 'ポチ')
animal_data = a.get_data()
print(animal_data)
````

関数と同様に，`return` で値を返せることが確認できた．

```{admonition} 画面に表示する print と，値を返す return
:class: tip

`print_data` は情報を画面に表示するメソッドである．一方，`get_data` は情報を呼び出し元に返すメソッドであり，返された値を変数に代入して，後の処理に利用できる．上の例でリストを表示しているのは，メソッドの外に書いた `print(animal_data)` である．
```

インスタンスメソッドには引数も与えることができる．以下では，`print_data` メソッドの引数に年齢を示す `age` を加えて，動物種や名前とともに表示する．


````python
class Animal:
    def __init__(self, species, name):
        self.species = species
        self.name = name

    def print_data(self, age):
        print(f'{self.species}の{self.name}は{age}歳です．')

    def get_data(self):
        animal_data = [self.species, self.name]
        return animal_data
````

まずインスタンスを作成し，次に `print_data` を呼び出すときに年齢を渡す．以下の例では，`a` が `self` に，`5` が `age` に渡される．これにより，メソッド内でインスタンス変数と引数の両方を使って処理できる．


````python
a = Animal('犬', 'ポチ')
a.print_data(5)
````

この例の `age` は，メソッドを呼び出すときに渡し，表示に使う引数である．`self.age` に代入していないため，年齢をインスタンス変数として保存しているわけではない．

以上より，クラスにデータだけでなく，そのデータを扱う機能（メソッド）を追加する方法を学んだ．情報の表示や取り出し方をクラスにまとめることで，同じクラスから作成した複数のインスタンスに対して，共通の処理を利用できる．


```{admonition} リストで返す方法と，タプルで返す方法
:class: tip

上記の `get_data` は，動物種と名前を一つのリストにまとめて返している．[第8回 複数の戻り値](#multiple-return-values)で学んだように，`return self.species, self.name` と書き，タプルとして返す方法もある．
```


### インスタンスメソッドをクラス内で呼び出す
インスタンスメソッドは，クラス内の別のメソッドから呼び出すこともできる．インスタンス変数を参照するときと同じように `self` を使い，`self.インスタンスメソッド名(引数)` と書くことで，同じインスタンスのメソッドを呼び出せる．以下のコードを参考にされたい．


````python
class Animal:
    def __init__(self, species, name):
        self.species = species
        self.name = name

    def print_data(self):
        print(f'{self.species}の{self.name}です．')

    def get_data(self):
        animal_data = [self.species, self.name]
        return animal_data

    def check_data(self):
        animal_data = self.get_data()
        print(animal_data)
````

上記の `Animal` クラスには，`get_data` を呼び出す `check_data` メソッドが追加されている．`check_data` では，`self.get_data()` でそのインスタンスの動物種と名前をリストとして受け取り，`print()` で表示している．

メソッド内から別のメソッドを呼び出せることを確認するために，インスタンスを作成し，`check_data` メソッドを呼び出す．


````python
a = Animal('犬', 'ポチ')
a.check_data()
````

`a.check_data()` を呼び出すと，`self` は `a` を示す．そのため，内部の `self.get_data()` も `a` の情報を取り出し，前述の例と同じ `['犬', 'ポチ']` が表示される．このように，既に定義したメソッドを利用すると，同じ処理を繰り返し書かずに済む．


### リストで学ぶインスタンスメソッド
「リストで学ぶインスタンス化」では，既に学んだ **リスト** も，リストの振る舞いを定義した `list` クラスのインスタンスであると説明した．リストは以下のように作成できた．


````python
numbers = [1, 2, 3]  # list クラスのインスタンスを作成
print(numbers)
````

ここで，再度，第2回の [リストの演算](#list-operations) にある要素の追加・削除・挿入の説明を読み返してほしい．例えば，要素を追加したいときは，`リスト.append(追加する値)` と呼び出すとあった．以下では，`numbers` の末尾に `100` を追加する．


````python
numbers.append(100)
print(numbers)
````

実は，この `append` もインスタンスメソッドなのである．`numbers.append(100)` は，リストのインスタンス `numbers` に対して，値を追加するメソッドを呼び出している．前述の `a.print_data(5)` と同じように，`インスタンス.メソッド名(引数)` という形で利用できる．

本講義では，クラス・インスタンス変数・インスタンスメソッドなど聞き慣れない用語が多数現れるが，リストを使う中で，既にインスタンスやメソッドを利用していたのである．各データ型が持つ機能を調べ，実行して確かめながら使い方に慣れてほしい．そして，クラスの概念と使い方をある程度理解できれば，Pythonのリストの機能を拡張したり，データ型自体を定義したりすることにもつながる．


```{admonition} クラスのまとめ

- クラスからデータの実体を作成することをインスタンス化といい，作成された実体をインスタンスという．`a = Animal('犬', 'ポチ')` では，作成したインスタンスを `a` に代入している．
```

## 【発展】 リストによるインスタンスの管理
これまでの例では，複数の動物の情報を表示するとき，`a.print_data()`，`b.print_data()` のように，頭数分の呼び出しを書いていた．インスタンスをリストにまとめると，ループを使って順番に処理できる．

````python
zoo = [
    Animal('犬', 'ポチ'),
    Animal('猫', 'タマ'),
    Animal('鳥', 'ピヨ')
]

for animal in zoo:
    animal.print_data()
````

```{admonition} リストとクラスを組み合わせて使う
:class: tip

```

もちろん，リストだけでなく辞書などの他のデータ型で管理しても良い．インスタンスを管理するクラスも考えることができる．管理する方法は作成したいアプリケーションの要件に依存するが，色々な実装を試してみると良い．


## 【発展】 クラスの継承

![](figs/inheritance.png)

これまでは新たな変数や機能を追加するために，`Animal` クラスを少しずつ書き換えてきた．Pythonを含むオブジェクト指向型言語には，クラスの機能を別のクラスで再利用する **継承** という仕組みが用意されている．ここでは，元のクラスを残したまま，その機能を受け継ぐ新たなクラスを定義する方法を説明する．

継承の方法は以下である．`class クラス名(継承元のクラス名):` と書くと，継承元のクラスで定義したメソッドを利用できる．この例の `super().__init__(variable)` は，継承元の初期化メソッドを呼び出している．

```python
class クラス名(継承元のクラス名):
    def __init__(self, variable, new_variable):
        super().__init__(variable)
        self.new_variable = new_variable

    def 新たに追加するインスタンスメソッド名(self, 引数):
        処理
```

この例のように，継承先で `__init__` を定義した場合，継承元の `__init__` は自動的には実行されない．継承元で行っている初期化処理も必要なら，`super().__init__(...)` で呼び出す．

```{admonition} 親クラスと子クラス
:class: tip

継承元のクラスは親クラスやスーパークラス，継承先のクラスは子クラスやサブクラスと呼ばれる．
```


実際に，継承を使って鳴き声の機能を持つ動物種ごとのクラスを作成してみる．これまでの `Animal` クラスは，動物種を示す `species` 変数を持っていた．ここでは，動物種ごとに異なる鳴き声を扱う方法として，犬や猫のクラスをそれぞれ定義する．

まずは，動物種を引数として受け取っていた `Animal` クラスを，継承の例に合わせて再定義する．名前の保存と表示を共通の機能とし，鳴き声の具体的な処理は子クラスで定義する．


````python
class Animal:
    def __init__(self, name):
        self.name = name
        
    def print_data(self):
        print(f'{self.name}です．')

    def sound(self):
        pass
````

クラスの設計段階で，処理をまだ記述しないメソッドには，上記のように **何もしない文** である `pass` を書くことができる．この例の `Animal.sound()` は，呼び出しても何も表示しない．具体的な鳴き声は，後で子クラスの `sound` メソッドに記述する．

```{admonition} メソッドの実装忘れに気づくには
:class: tip

`pass` の代わりに `raise NotImplementedError` と書く方法もある．子クラスで `sound` を定義し忘れた場合，そのメソッドが呼び出されたときにエラーが発生するため，実装忘れに気づきやすくなる．例外処理の詳細は本講義では扱わないが，興味がある受講生は [Pythonドキュメント](https://docs.python.org/ja/3/tutorial/errors.html) などを参考にされたい．
```

では，新たに定義した `Animal` クラスを継承し，犬クラス（`Dog`）と猫クラス（`Cat`）を定義する．


````python
class Dog(Animal):  # 継承元のクラスを指定
    def __init__(self, name):
        super().__init__(name)  # 親クラスの初期化メソッドを呼び出す

    def sound(self):  # 犬の鳴き声メソッド
        print('ワンワン')


class Cat(Animal):
    def __init__(self, name):
        super().__init__(name)

    def sound(self):  # 猫の鳴き声メソッド
        print('ニャア')
````

まず，継承の利点を理解するために，`Dog` と `Cat` のインスタンスで，名前を示す `name` と，親クラスのメソッド `print_data` を利用できることを確認されたい．`name` は，`super().__init__(name)` によって親クラスの初期化処理を呼び出すことで，それぞれのインスタンスに保存される．


````python
zoo = [Dog('ポチ'), Cat('タマ')]

for animal in zoo:
    animal.print_data()
````

このように，名前を保存する処理や `print_data` を子クラスに重ねて書かずに，親クラスの機能を利用できることが継承の利点である．

続いて，`Dog` と `Cat` で定義し直したメソッド `sound` を呼び出す．


````python
zoo = [Dog('ポチ'), Cat('タマ')]

for animal in zoo:
    animal.print_data()
    animal.sound()
    print('') # 改行用のprint
````

以上より，親クラスから受け継いだ `print_data` と，子クラスで定義し直した `sound` の両方を呼び出せることが確認できた．同じ `animal.sound()` という呼び出しでも，犬なら「ワンワン」，猫なら「ニャア」と，取り出したインスタンスに応じた処理が実行される．

```{admonition} オーバーライド
:class: tip

`sound` メソッドのように，親クラスと同じ名前のメソッドを子クラスで定義し，処理を置き換えることを **オーバーライド** と呼ぶ．この例では，親クラスの `print_data` はそのまま利用し，`sound` の処理だけを動物種に合わせて変更している．
```

共通の処理を親クラスにまとめておけば，修正する箇所をまとめやすくなる．ただし，ノートブックで親クラスを再定義しただけでは，既に定義した子クラスや作成済みのインスタンスは更新されない．変更を確認するときは，親クラス，子クラス，インスタンス作成の順にセルを実行されたい．


補足となるが，このようなクラス設計をしておくと，名前が違う別の犬を管理するときも，同じクラスを利用できる．以下のサンプルコードを確認されたい．


````python
dog = Dog('ポチ')
another_dog = Dog('クロ')

dog.print_data()
another_dog.print_data()
````

このように，同じ `Dog` クラスから作成しても，ポチとクロはそれぞれ別のインスタンスとして名前を持つ．同一種の動物で共通の処理を利用しながら，個体ごとに異なるデータを管理できる．複数の犬をまとめて扱いたい場合は，前述のようにインスタンスをリストに入れて管理しても良い．
