## Google Colaboratoryの使い方

本講義では，Google Colaboratoryを用いてPythonによるプログラミングを学びます．
Google Colaboratoryとは，その名の通りGoogleが提供するJupyter Notebook環境であり，クラウド上で実行できるため，ブラウザがあれば無料で利用することができます．

Googleアカウントを持っていない人は，あらかじめ作成しておきましょう．

### Colaboratoryの立ち上げ

ブラウザにGoogleアカウントでログインした状態で，以下のURLを開いてください．

- [https://colab.research.google.com](https://colab.research.google.com)

以下の画面が出たら，ひとまず「キャンセル」を選択しておいてください．

## Google Colaboratoryを使った教材の開き方

本講義で利用する教材は，サポートページから直接Colaboratoryで開くことができます．

実際に，このページをColaboratoryで開いてみましょう．

サポートページのトップにある「講義資料」の表で，開きたい教材の行の「Colab」列にある「Open」をクリックしてください．

すると，以下のようにGoogle Colaboratory上で本教材が開かれます．

![](./figs/colab_launch.png)

Google Colaboratoryで開いた教材を用いることで，実際に教材中のプログラムを動かしたり，自分でちょっと書き換えたりすることができます．

````{warning}
上記の手順で開いた教材は，そのままではノートブックを操作した結果が **Googleドライブ上に保存されません**．
ブラウザのタブを閉じると，書き換えた内容や実行結果は失われてしまいます．

ノートブックの上側にある「**ドライブにコピー**」というボタンを押すと，自分のGoogleドライブ上にノートブックのコピーが作成されます（`マイドライブ -> Colab Notebooks` フォルダ内）．

````

```{note}
Google Colaboratoryでは，一定時間操作しなかったり，長時間使い続けたりすると，**ランタイム**（プログラムを実行している環境）が自動的に切断されます．
切断されるとノートブック自体は残りますが，それまでに実行した結果（後の回で学ぶ変数の値など）はリセットされます．

その場合は，メニューの「ランタイム」→「すべてのセルを実行」を選ぶと，上から順にすべてのコードセルを実行し直すことができます．
```

### 簡単な算術計算

さて，文字列を表示する以外の例として，算術計算もやってみましょう．

コードセル内に計算式を記入して実行すると，電卓のようにその計算結果が出力されます．

`````python
1 + 1
`````

`1 + 1` の結果である `2` が出力されたかと思います．

### コメント

Pythonでは，コード中に `#` が記載してあると，それ以降，その行の終わりまでが **コメント** として扱われます．
コメントの部分はプログラム実行時には無視されるため，コードに関する補足説明などを記載するときに利用されます．

`````python
# このように行頭に `#` を書くことで，その行全体がコメントとなる．

10 / 5 # このように行の途中からコメントを書くこともできる
`````

`````python
# 以下のようにコード行にコメントを付け，その部分を無視させることもよく行われる．

# 10 / 5
10 / 2 # こちらのみ実行される
`````

今後，コード中に補足説明などを示す場合にコメントが登場するので，注意しておきましょう．
なお，コメントの部分は色も変わるので，実行されるコード部分との違いは比較的分かりやすいかと思います．

## 【コラム】生成AIとの付き合い方

ChatGPT や Gemini などの生成AIは，Pythonのコードを書いたり，エラーの原因を説明したりすることが得意です．
Google Colaboratory にも Gemini が組み込まれており，コードの自動補完などを利用できます．

とても便利な道具ですが，課題の答えを丸ごと出してもらい，理解しないまま写すだけでは，プログラミングの力は身につきません．
生成AIの出力にはもっともらしい誤りが含まれることもあり，それに気づいて直せるのは，基礎を理解している人だけです．

生成AIは「答えを出してくれる機械」ではなく，**いつでも質問できる家庭教師** として使うのがおすすめです．
例えば，エラーメッセージの意味を聞く，教材のわかりにくい部分を別の例で説明してもらう，自分で解き終わった後に別解を聞いて比べる，といった使い方です．
