```{admonition} 変数定義のルール

数学で学ぶ $=$ は右辺と左辺が **等しい** ことを，プログラミングの `=` は右側の値を左側の変数に **代入する** ことを示す．意味が異なることに注意されたい．そのため，`x = 10` の左右を入れ替えて `10 = x` と書くことはできない．
```

### 式を計算する
前述した式 $y=2x+1$ の右辺を，変数 `x` に代入した値 `10` を使って計算する．第1回で学んだように，Pythonでは数値同士の四則演算がサポートされている．足し算は `+` 演算子，掛け算は `*` 演算子で実行することができる．数値を直接書く代わりに変数名を書くと，その変数に代入した値を使って計算できる．

````python
2 * x + 1
````

````python
print(y)
````

### 代入した値を変更する
![](figs/reassign.png)

では，最後に，変数 `x` の値を変更することを試みる．上記の代入手順を再度行うことで，変数が参照する値を新たなデータで更新できる．この操作を **再代入** と呼ぶ．以下のセルでは `x` に `100` を代入し，値が変わることを確認する．

````python
x = 100
````

````python
print(x)
````

`x` の値は `100` に変わった．では，先ほど `y = 2 * x + 1` で計算した `y` の値も変わっただろうか？ 次のセルで確認されたい．

````python
print(y)
````

`y` は `21` のままである．`y` には，代入した時点の **計算結果** が保存されているため，後から `x` の値を変えても，自動では再計算されない．

新しい値での計算結果を得るには，もう一度式を計算して `y` に代入する必要がある．再代入した変数 `x` で，前述の式を計算してみる．

````python
y = 2 * x + 1
print(y)
````

結果が `21` から `201` に変わったことがわかる．$y=2x+1$ の $x$ に値を代入して式を計算するシンプルな問題を通して，変数と演算のイメージを掴んでいただきたい．

```{admonition} 代入・参照・更新のまとめ
- 同じ変数に別の値を代入すると，その変数が参照する値を更新できる．
- 計算に使った変数の値を変えても，過去の計算結果は自動では更新されない．必要な式をもう一度実行する．
```

## 変数の型と名前
先ほどの例では，数値を扱う変数を定義して演算を行った．Pythonでは数値だけでなく文字列，論理値，複数の値をまとめたデータなど，様々なデータを変数に代入して扱うことができる．このようなデータの種類を **データ型** （または単に，**型**）と呼ぶ．Pythonによってサポートされるデータ型は多岐にわたる．本講義で取り上げる代表的なデータ型を詳しく紹介する前に，型の確認方法と変数の名前について説明する．

以下に，発展内容も含めて，本講義で扱う基本的なデータ型をまとめる．

| データ型の名前 | 型 | 例 |
| --- | --- | --- |
| 整数 | `int` | `42` |
| 浮動小数点数 | `float` | `3.14` |
| 複素数 | `complex` | `1 + 2j` |
| 文字列 | `str` | `"Hello, World!"` |
| リスト | `list` | `[1, 2, 3]` |
| タプル | `tuple` | `(1, 2, 3)` |
| 辞書 | `dict` | `{'key': 1}` |
| ブール値 | `bool` | `True`, `False` |

```{admonition} 【発展】動的型付け
:class: tip

Pythonでは，変数を定義するときに，データの型をあらかじめ指定する必要はない．例えば，`x = 4` とすると `x` は `int` 型のデータを参照し，その後で `x = 'Hello'` とすると `str` 型のデータを参照する．このように，変数に特定の型を固定せず，実行時に扱うデータの型に応じて処理を行う仕組みを **動的型付け** と呼ぶ．

型をあらかじめ指定することなく変数を利用できる利点を持つが，大規模なプログラムを作成したり，変数が増えたりすると，意図しない型のデータ同士で演算をしてしまうことがある．このような型の取り違えは，プログラムが意図どおりに動かない原因（**バグ**）となる．バグを生じさせないためにも，変数に代入したデータの型を把握しておくことが重要である．
```

### 変数の型の確認方法
変数の型を調べるときは，以下で確認できる．ここで「変数の型」とは，その変数が参照しているデータの型のことである．

```python
type(変数名)
```

以下のコードでは整数 `4` を代入した変数 `x` の型を確認している．Pythonでは整数を扱うデータ型は `int` 型と呼ばれ，`type(x)` の出力が `int` であることが確認できる．

````python
x = 4
type(x)
````

```{admonition} 組み込み関数

`print()` や `type()` など，Pythonがあらかじめ提供する関数を **組み込み関数** と呼ぶ．Pythonでは様々な組み込み関数が提供されており，Pythonの[ドキュメントページ](https://docs.python.org/ja/3/library/functions.html)で確認できる．
```

### 変数名

データ型に加えて，変数の名前にも注意が必要である．変数名には小文字の英字（`a-z`），大文字の英字（`A-Z`），数字（`0-9`），アンダースコア（`_`）を利用することができる．ただし，先頭に数字を置いたり，名前に空白やハイフン（`-`）を含めたりすることはできない．何らかのエラーが生じた際には，変数の名前も確認されたい．

以下に，変数名の例を示す．

| 変数名 | 利用できるか | 説明 |
| --- | --- | --- |
| `student_name` | できる | 英字とアンダースコアを使っている． |
| `score1` | できる | 先頭以外には数字を使える． |
| `1score` | できない | 先頭に数字を使っている． |
| `student name` | できない | 空白を含んでいる． |
| `student-name` | できない | ハイフンを含んでいる． |

Pythonでは日本語なども変数名に利用できるが，本講義では英数字とアンダースコアを基本とする．変数名は，`student_name` のように扱うデータの意味がわかる名前にすると，後からプログラムを読み返しやすい．通常は小文字の英字を使い，複数の単語はアンダースコアで区切る．

また，変数名では **大文字と小文字が区別される**．以下のコードを実行し，`score` と `Score` が別の変数として扱われることを確認されたい．

````python
score = 80
Score = 90
print(score)
print(Score)
````

**変数名に使えない文字列**

以下の単語は，**予約語** と呼ばれる，プログラム上で特別な役割を持った単語のため，変数名には利用できない．

```text
False, None, True, and, as, assert, async, await, break,
class, continue, def, del, elif, else, except, finally,
for, from, global, if, import, in, is, lambda, nonlocal,
not, or, pass, raise, return, try, while, with, yield
```

```{admonition} 使えない名前と避けたい名前
:class: caution

予約語と，組み込み関数などの名前は区別する必要がある．

- `class` や `for` などの予約語は，変数名に使うとエラーになる．
- `print`，`type`，`list` などの名前は予約語ではないため，変数名として使えてしまう．しかし，例えば `print = 10` と代入すると，その名前で元の `print()` 関数を呼び出せなくなる．このような混乱を避けるため，組み込み関数などの名前を変数名として使わないようにされたい．
```

```{admonition} 変数のまとめ

- 変数の仕組みによってコンピュータ内部の挙動を把握することなくデータの読み書きができる．
- 変数に代入するデータにはいくつかの種類があり，この種類のことをデータの型という．
- データの型は `type()` で知ることができる．
- 変数名の付け方にはルールがあり，大文字と小文字も区別される．
```

### 数値データの演算
数値データの定義の次は，数値データ同士の四則演算を紹介する．Pythonでは，`int` 型や `float` 型といった数値データに対して，一般的な四則演算がサポートされている．第1回の導入で行った数値データの演算を，変数を使って同様に行うことができる．

数値データを代入した変数 `a` と `b` を作成し，足し算 `+`，引き算 `-`，掛け算 `*`，割り算 `/`，切り捨て除算 `//`，冪乗（べき乗）`**`，剰余（余り）`%` の計算例を確認する．

````python
a = 11
b = 3
print(a)
print(b)
````

````python
a + b
````

````python
a - b
````

````python
a * b
````

````python
a / b
````

```{admonition} 割り算の結果の型
:class: tip

`/` を使った割り算の結果は，整数同士で計算しても `float` 型となる．割り切れる場合も同様で，例えば `6 / 3` の結果は `2.0` となる．
```

````python
print(6 / 3)
print(type(6 / 3))
````

`//` を使うと，割り算の結果を，それ以下の最も近い整数にできる．例えば，`11 / 3` は約 `3.67` なので，`11 // 3` の結果は `3` となる．この演算を **切り捨て除算** と呼ぶ．

````python
a // b
````

```{admonition} 負の数の切り捨て除算
:class: caution

負の数では，小数部分を単に取り除いた結果とは異なることに注意されたい．例えば，`-11 / 3` は約 `-3.67` である．`-3` はこれより大きいため，`-11 // 3` の結果は，それ以下で最も近い整数の `-4` となる．以下のコードで，正の数と負の数の結果を比べてみよう．

なお，`//` の結果が常に `int` 型になるわけではない．例えば，`11.0 // 3` のように `float` 型を使って計算すると，結果も `3.0` のような `float` 型になる．
```

````python
print(11 // 3)
print(-11 // 3)
````

冪乗（べき乗）の計算は `**` で実行できる．例えば，$a^{b}$ は `a ** b` と書く．

````python
a ** b
````

割り算の余り（剰余）は `%` で計算できる．例えば，`11` を `3` で割ると商が `3`，余りが `2` なので，`11 % 3` の結果は `2` となる．

````python
a % b
````

複数の演算を一つの式にまとめることができる．

````python
a + 2 * b / 3.14
````

数式の計算と同じように，演算には優先順位がある．例えば，掛け算や割り算は足し算や引き算より先に計算される．先ほどの式では `2 * b / 3.14` が先に計算され，その結果に `a` が足される．一方，`()` で囲んだ部分は優先して計算されるため，以下の式では `a + 2` が先に計算される．

````python
(a + 2) * b / 3.14
````

```{admonition} 小数の計算には誤差が生じることがある
:class: tip
```

#### 文字列中の文字を参照する

![](figs/slice.png)

続いて，文字列中の文字を参照する方法を紹介する．例えば，`Hello` から最後の文字 `o` を取り出す場面である．このような場面では，文字の位置を表す **インデックス（index）** を指定することで参照できる．インデックスは，文字列の各文字に付けられた番号であることを上図からイメージされたい．参照したい文字のインデックスがわかったら，以下のように指定する．

```python
文字列[インデックス]
```

注意点として，**先頭の文字のインデックスは `0`** である．そのため，`0` から順に数えて，参照したい文字のインデックスを求める必要がある．また，インデックスを指定して文字を参照する処理を **インデキシング（indexing）** と呼ぶ．

変数 `s` に文字列 `Hello` を代入し，各文字を参照する．`s[0]` で先頭の `H`，`s[4]` で最後の `o` を取り出すことができる．

````python
s = 'Hello'
s[0]
````

````python
s[4]
````

`len()` 関数で文字数を取得できるため，`len(s) - 1` とすると最後の文字のインデックスが得られる．以下の例では，`len(s)` が `5` なので，`s[4]` と同じ文字を参照する．

````python
s[len(s) - 1]
````

負のインデックスを使うと，末尾から数えて文字を参照できる．末尾の文字は `-1`，その一つ前は `-2` となる．`Hello` の各文字とインデックスの対応を以下に示す．

| 文字 | `H` | `e` | `l` | `l` | `o` |
| --- | --- | --- | --- | --- | --- |
| 先頭から数えるインデックス | `0` | `1` | `2` | `3` | `4` |
| 末尾から数えるインデックス | `-5` | `-4` | `-3` | `-2` | `-1` |

````python
s[-1]
````

```{admonition} 存在しない位置は参照できない
:class: caution

`Hello` は5文字なので，先頭から数えるインデックスは `0` から `4` までである．`s[5]` のように，存在しない位置を指定すると `IndexError` になる．末尾から数える場合も，この文字列では `-5` まで指定できるが，`s[-6]` はエラーになる．
```

#### 文字列中の部分文字列を参照する

![](figs/slice_start_end.png)

前述のように，文字列から1文字を取り出す際はインデックスを指定すれば良い．ここでは，文字列の一部分である **部分文字列** を参照することを考える．例えば，上図のように文字列 `Hello` から部分文字列 `ell` を取り出す場面である．このような処理を実現するためには，参照したい部分文字列の開始インデックスと終了インデックスを，コロン `:` を用いて以下のように指定すれば良い．

```python
文字列[開始インデックス:終了インデックス]
```

**開始インデックスの文字は含まれるが，終了インデックスの文字は含まれない**．文字列から部分文字列を上記のように参照することを **スライシング（slicing）** と呼ぶ．

スライシングを利用せずに `ell` を取り出すには，以下のように各文字を一つずつ参照して連結する方法がある．しかし，取り出したい文字数が増えると，可読性が低い冗長なプログラムになる．

````python
s = 'Hello'
````

````python
print(s[1] + s[2] + s[3])
````

スライシングを利用すると，同じ部分文字列をシンプルに参照できる．

````python
s[1:4]
````

```{admonition} スライシングの終了位置
:class: caution

`s[1:4]` では，インデックスが `1`，`2`，`3` の文字が取り出される．終了インデックスに指定した `4` の文字 `o` は含まれないことに注意されたい．この例のように左から右へ続けて文字を取り出す場合，終了インデックスには「取り出したい最後の文字のインデックス + 1」を指定すると良い．

また，インデキシングとは異なり，スライシングでは終了位置が文字列の長さを超えてもエラーにはならない．例えば，`s[:100]` は末尾までの文字列を取り出す．
```

次のようなスライシングも可能である．ここからは，文字列 `Hello World!` を使って確認する．

````python
s = 'Hello World!'
s
````

あるインデックス $i$ から文字列の末尾までを部分文字列として参照する際は，終了インデックスを省略できる．

```python
文字列[開始インデックス:]
```

````python
i = 3
s[i:]
````

`len()` 関数で取得できる文字数は，最後の文字のインデックスより1大きい．そのため，上記の処理は，終了インデックスに `len(s)` を指定する以下の処理と等しい．

````python
s[i:len(s)]
````

文字列の先頭（インデックス `0`）から，あるインデックス $j$ の手前までを部分文字列として参照する際は，開始インデックスを省略できる．この場合も，インデックス $j$ の文字は含まれない．

```python
文字列[:終了インデックス]
```

````python
j = 3
s[:j]
````

先頭の文字のインデックスは `0` であるため，上記の処理は `文字列[0:終了インデックス]` と等しい．

````python
s[0:j]
````

開始インデックス $i$ から終了インデックス $j$ の手前まで，インデックスを $k$ ずつ進めて参照する場合は，以下のように書ける．この $k$ を **ステップ** と呼ぶ．

```python
文字列[開始インデックス:終了インデックス:ステップ]
```

ステップが `1` なら続けて，`2` なら一文字おきに取り出す．以下の例では，インデックスが `2`，`4`，`6` の文字を取り出すため，結果は `'loW'` となる．ステップを省略した場合は `1` として扱われる．なお，ステップに `0` は指定できない．

````python
i, j, k = 2, 8, 2
s[i:j:k]
````

```{admonition} 開始位置と終了位置を両方省略する
:class: tip

`s[:]` とすると文字列全体を取り出すことができる．また，`s[::2]` とすると，先頭から一文字おきに取り出せる．

さらに，ステップに負の数を指定すると，逆向きに文字を取り出すことができる．例えば，`s[::-1]` は文字列全体を末尾から逆順に取り出す．以下のコードで，結果を比べてみよう．
```

````python
print(s[:])
print(s[::2])
print(s[::-1])
````

```{admonition} 文字列の一部は直接書き換えられない
:class: caution

インデキシングで文字を参照することはできるが，`s[0] = 'h'` のように，文字列の一部に別の文字を代入することはできない．このような代入をすると `TypeError` になる．

文字列を変更したい場合は，新しい文字列を作って変数に代入する．以下の例では，文字列 `Hello` の先頭を小文字にするため，`'h'` と2文字目以降の部分文字列を連結し，変数 `greeting` に代入し直している．
```

````python
greeting = 'Hello'
greeting = 'h' + greeting[1:]
print(greeting)
````

```{admonition} 文字列データ型のまとめ

- 文字列は `str` 型として扱われ，クォーテーションで囲んで書く．
- 文字列の長さは `len()` 関数で取得できる．空白や改行も文字数に含まれる．
- インデキシングで1文字を参照できる．先頭は `0`，末尾は `-1` で指定できる．
- スライシングで部分文字列を参照できる．開始位置は含まれるが，終了位置は含まれない．
- 文字列の一部を直接書き換えることはできない．変更したい場合は，新しい文字列を作って変数に代入する．
```

#### リストの連結


`+` は連結した新しいリストを作るため，元の二つのリストは変更されない．以下の例では，連結したリストを変数 `l3` に代入している．

````python
l1 = [1, 2, 3]
l2 = [10, 11, 12]
l3 = l1 + l2
````

````python
print(l3)
````

3つ以上のリストの連結も同時にできる．

````python
l1 = [1, 2, 3]
l2 = [10, 11, 12]
l3 = [100, 101, 102]
l4 = l1 + l2 + l3
````

````python
print(l4)
````

```{admonition} append()とextend()の違い
:class: caution

二つのリストを連結したいときに，`append()` メソッドを利用すると，意図した結果にならないことがある．追加の仕方に違いがあるためである．

- `l1.append(l2)` は，`l2` というリスト全体を，一つの要素として `l1` の末尾に追加する．
- `l1.extend(l2)` は，`l2` の各要素を，順番に `l1` の末尾に追加する．

どちらのメソッドも，`l1` 自体を変更する．以下の二つの例で，結果の違いを確認されたい．まずは `append()` を使った例を示す．
```

## 【発展】データ型の変換

Pythonでは，関数を使って値を別のデータ型に変換することができる．このような変換を **型変換**，または **キャスト** と呼ぶ．数値を文字列として表示したい場合や，文字列で与えられた数値を計算に使いたい場合などに便利な仕組みである．ただし，どの値でも任意のデータ型に変換できるわけではない．

文字列の例で確認したように，出力が数値に見えても，データ型が数値データ型であるとは限らない．また，文字列と数値データでは，同じ演算子でも処理が異なる．このようなデータ型と演算子による処理の違いはバグの原因となりうるため，プログラムを作成する際は，`type()` 関数でデータ型を確認することを心がけたい．以降，主要なデータ型について，変換が必要な場面とその変換方法を紹介する．

### `int`型から`float`型へ，`float`型から`int`型への変換
`int` 型から `float` 型への変換は `float()` 関数を，`float` 型から `int` 型への変換は `int()` 関数を利用することで行うことができる．変換後の値を変数に代入して使う．

ただし，`float` 型の値を `int` 型に変換すると，小数部分を取り除いた整数になる．四捨五入ではないことに注意されたい．以下に変換例を示す．

````python
x = 2
print(x)
print(type(x))
````

````python
x = float(x)
print(x)
print(type(x))
````

````python
x = 3.14
print(x)
print(type(x))
````

````python
x = int(x)
print(x)
print(type(x))
````

```{admonition} 負の数をint型に変換する場合
:class: caution

負の数でも，小数部分を取り除く．そのため，`int(-3.14)` は `-3` となる．

前半で学んだ `//` とは結果が異なることに注意されたい．`-3.14 // 1` は，割り算の結果以下で最も近い整数を求めるため，`-4.0` となる．
```

````python
print(int(3.14))
print(int(-3.14))
print(-3.14 // 1)
````

### タプル型からリスト型へ，リスト型からタプル型への変換
タプル型（`tuple`）をリスト型（`list`）へ，またリスト型からタプル型へ変換する方法を学ぶ．タプルとリストの違いは，要素の変更ができるかどうかである．プログラム内で要素を意図せず書き換えることを避けたいときは，リストからタプルへ変換すると良い．また，ユーザ入力に応じた変更など，要素を書き換えたい場面では，タプルからリストへ変換して扱うことができる．

#### タプル型からリスト型への変換

````python
data = ('a', 'b', 'c')
print(data)
print(type(data))
````

タプルは要素を書き換えられない．以下のコードを実行すると，`TypeError` が発生することを確認されたい．

````python
data[0] = 'd'
````

`list(data)` とすると，タプル `data` の要素を並べた新しいリストを作ることができる．以下では，作成したリストを `list_data` に代入し，先頭の要素を変更する．元のタプル `data` の要素は変わらない．

````python
list_data = list(data)
print(list_data)
print(type(list_data))
````

````python
list_data[0] = 'd'
print(list_data)
````

#### リスト型からタプル型への変換
`tuple(list_data)` とすると，リスト `list_data` の要素を並べた新しいタプルを作ることができる．

````python
tuple_data = tuple(list_data)
print(tuple_data)
print(type(tuple_data))
````

変換後のタプルも要素を書き換えられない．以下のコードを実行すると，`TypeError` が発生することを確認されたい．

````python
tuple_data[0] = 'e'
````

```{admonition} データ型の変換のまとめ

- `int()`，`float()`，`str()`，`list()`，`tuple()` などを使って，値を別のデータ型に変換できる．
- 変換できるかどうかは，元の値と変換先の型によって決まる．変換できない値を渡すと，エラーになる．
- 表示が同じように見えても，型が異なると演算の結果が変わることがある．迷ったときは `type()` 関数で確認する．
```
